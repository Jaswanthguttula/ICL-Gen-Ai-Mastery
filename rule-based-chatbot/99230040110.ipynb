```python
name = input("What is your name: ").lower().strip()

while True:
    message = input("You: ").lower().strip()

    if "exit" in message or "bye" in message:
        print(f"Bye {name}")
        break

    elif "hello" in message or "hi" in message:
        print(f"Hey there {name}")

    elif "joke" in message:
        print("That's great. Take a spoon and fill it with water.")
        print("Then jump into that 😅")

    elif "how are you" in message:
        print(f"Thanks for asking....💗 How are you Mr. {name}?")

    else:
        print("Sorry I can't hear you. Try with another message")
```
